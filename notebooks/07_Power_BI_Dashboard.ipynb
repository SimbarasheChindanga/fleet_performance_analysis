```python
# CELL 70: Export data for Power BI

import pandas as pd
from pathlib import Path

BASE_PATH = Path.cwd().parent
PROCESSED_DATA_PATH = BASE_PATH / 'data' / 'processed'
POWERBI_DATA_PATH = BASE_PATH / 'powerbi' / 'data'

# Create powerbi/data folder
POWERBI_DATA_PATH.mkdir(parents=True, exist_ok=True)

print("=" * 60)
print("📊 EXPORTING DATA FOR POWER BI")
print("=" * 60)

print(f"📍 Data will be saved to: {POWERBI_DATA_PATH}")

# Load cleaned data
trucks_df = pd.read_csv(PROCESSED_DATA_PATH / 'trucks_clean.csv')
drivers_df = pd.read_csv(PROCESSED_DATA_PATH / 'drivers_clean.csv')
trips_df = pd.read_csv(PROCESSED_DATA_PATH / 'trips_clean.csv')
maintenance_df = pd.read_csv(PROCESSED_DATA_PATH / 'maintenance_clean.csv')
fuel_df = pd.read_csv(PROCESSED_DATA_PATH / 'fuel_purchases_clean.csv')

# 1. Trip Summary - Main fact table
trip_summary = trips_df.copy()
trip_summary.to_csv(POWERBI_DATA_PATH / 'trip_summary.csv', index=False)
print(f"✅ Trip Summary: {len(trip_summary)} rows")

# 2. Truck Summary - Dimension table
truck_summary = trucks_df.copy()
truck_summary.to_csv(POWERBI_DATA_PATH / 'truck_summary.csv', index=False)
print(f"✅ Truck Summary: {len(truck_summary)} rows")

# 3. Driver Summary - Dimension table
driver_summary = drivers_df.copy()
driver_summary.to_csv(POWERBI_DATA_PATH / 'driver_summary.csv', index=False)
print(f"✅ Driver Summary: {len(driver_summary)} rows")

# 4. Route Summary - Aggregated route metrics
route_summary = trips_df.groupby('route').agg({
    'trip_id': 'count',
    'distance_km': 'sum',
    'revenue': 'sum',
    'profit': 'sum',
    'fuel_consumed_litres': 'sum',
    'fuel_efficiency_km_per_l': 'mean'
}).reset_index()
route_summary.columns = ['route', 'trip_count', 'total_distance_km', 'total_revenue', 
                         'total_profit', 'total_fuel_litres', 'avg_fuel_efficiency']
route_summary.to_csv(POWERBI_DATA_PATH / 'route_summary.csv', index=False)
print(f"✅ Route Summary: {len(route_summary)} rows")

# 5. Monthly Summary - Time series
monthly_summary = trips_df.groupby(['trip_year', 'trip_month']).agg({
    'trip_id': 'count',
    'distance_km': 'sum',
    'revenue': 'sum',
    'profit': 'sum',
    'fuel_consumed_litres': 'sum'
}).reset_index()
monthly_summary['month_year'] = monthly_summary['trip_year'].astype(str) + '-' + monthly_summary['trip_month'].astype(str).str.zfill(2)
monthly_summary.to_csv(POWERBI_DATA_PATH / 'monthly_summary.csv', index=False)
print(f"✅ Monthly Summary: {len(monthly_summary)} rows")

# 6. Maintenance Summary
maintenance_summary = maintenance_df.copy()
maintenance_summary.to_csv(POWERBI_DATA_PATH / 'maintenance_summary.csv', index=False)
print(f"✅ Maintenance Summary: {len(maintenance_summary)} rows")

# 7. Fuel Summary
fuel_summary = fuel_df.copy()
fuel_summary.to_csv(POWERBI_DATA_PATH / 'fuel_summary.csv', index=False)
print(f"✅ Fuel Summary: {len(fuel_summary)} rows")

print("\n" + "=" * 60)
print("✅ All data exported to Power BI folder!")
print(f"📁 Location: {POWERBI_DATA_PATH}")
print("\n📋 Files created:")
for file in POWERBI_DATA_PATH.glob('*.csv'):
    size_kb = file.stat().st_size / 1024
    print(f"   - {file.name} ({size_kb:.1f} KB)")
print("=" * 60)

print("\n📋 Next Steps:")
print("1. Open Power BI Desktop")
print("2. Click 'Get Data' → 'Text/CSV'")
print("3. Import each CSV file from the powerbi/data/ folder")
print("4. Create relationships between tables")
print("5. Build your dashboard!")
```
